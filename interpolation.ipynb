```python
import pandas as pd

data = {
    "name": ["Alice", "Shreechan", "Charlie", "David", "Eva", "Ragav", "Shaurav", "Rajiv", "Bibek", "Shanjiv"],
    "age": [25, None, 35, 40, 45, 29, 26, 26, 21, 23],
    "city": ["New York", None, "Chicago", "Houston", "Phoenix", "Dolkha", "Lamjung", "Tanahun", "Lamjung", "Kathmandu"],
    "blood_type": ["A", None, "O", "AB", "A", "B+", "B", "A", "A+", "O+"],
    "exam_marks": [23, None, 23, 11, 67, 43, 76, 23, 65, 22]
    
}
df = pd.DataFrame(data)
```

1. linear
2. polymonial
3. time

```python
df.interpolate(method="linear", axis= 0, inplace= True)
```

```python
df['exam_marks'] = df['exam_marks'].interpolate(method='polynomial', order=2)
```

```python
df['blood_type'].fillna(df['blood_type'].mode().iloc[0], inplace= True, axis = 0)
```

```python
df['city'].fillna(df['city'].mode().iloc[0], inplace=True, axis=0)
```

```python
df['age'] = df['age'].interpolate(method='linear')
```

```python
df.sort_values(by="name", ascending=True, inplace=True)
```
