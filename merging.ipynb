```python
import pandas as pd
```

```python
df_customer = pd.DataFrame({
    "customer_id" : [1,2,3],
    "Name": ["bibek", "bibek_don", "dami_bibek"]
})
```

```python
df_order = pd.DataFrame({
    "customer_id" : [1, 2, 4],
    "orderamount": [1000,32322, 69699]
})
```

```python
#merge = inner join
df_merged = pd.merge(df_customer, df_order, on = "customer_id", how = "inner")
print('inner join')
print(df_merged)
```

```python
#merge = outer join
df_merged = pd.merge(df_customer, df_order, on = "customer_id", how = "outer")
print('outer join')
print(df_merged)
```

```python
#merge = left join
df_merged = pd.merge(df_customer, df_order, on = "customer_id", how = "left")
print('left join')
print(df_merged)
```

```python
#merge = right join
df_merged = pd.merge(df_customer, df_order, on = "customer_id", how = "right")
print('right join')
print(df_merged)
```

```python
#merge = left join
df_merged = pd.merge(df_customer, df_order, on = "customer_id", how = "left")
print('left join')
print(df_merged)
```
